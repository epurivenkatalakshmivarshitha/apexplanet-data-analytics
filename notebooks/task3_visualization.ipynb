```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.io as pio

df = pd.read_csv('../data/cleaned_superstore.csv')
df['Order Date'] = pd.to_datetime(df['Order Date'])
df.head()
```

```python
# Line chart - sales trend
plt.figure(figsize=(10,5))
df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum().plot(kind='line')
plt.title('Monthly Sales Trend')
plt.savefig('../dashboards/monthly_sales_trend.png')
plt.show()
```

```python
# Bar chart - sales by category
plt.figure(figsize=(8,5))
df.groupby('Category')['Sales'].sum().plot(kind='bar', color='teal')
plt.title('Sales by Category')
plt.savefig('../dashboards/sales_by_category.png')
plt.show()
```

```python
# Scatter plot - Sales vs Profit
plt.figure(figsize=(8,5))
plt.scatter(df['Sales'], df['Profit'], alpha=0.4)
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.savefig('../dashboards/sales_vs_profit.png')
plt.show()
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(8,5))
df.groupby('Region')['Sales'].sum().plot(kind='bar', color='coral')
plt.title('Sales by Region')
plt.ylabel('Sales')
plt.savefig('../dashboards/sales_by_region.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Category', y='Sales', data=df)
plt.title('Sales Distribution by Category')
plt.savefig('../dashboards/sales_boxplot_category.png')
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby('Sub-Category')['Sales'].sum().sort_values().plot(kind='barh', color='steelblue')
plt.title('Sales by Sub-Category')
plt.xlabel('Sales')
plt.savefig('../dashboards/sales_by_subcategory.png')
plt.show()
```
